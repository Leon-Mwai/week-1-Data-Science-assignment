```python
import pandas as pd
import sqlite3


data = {
    "Name": ["Wii Sports", "Super Mario Bros", "Mario Kart Wii", "Pokemon Red", "Tetris", 
             "New Super Mario Bros", "Wii Sports Resort", "Pokemon Gold", "Gran Turismo", "Duck Hunt"],
    "Platform": ["Wii", "NES", "Wii", "GB", "GB", "DS", "Wii", "GB", "PS", "NES"],
    "Year": [2006, 1985, 2008, 1996, 1989, 2006, 2009, 1999, 1997, 1984],
    "Genre": ["Sports", "Platform", "Racing", "Role-Playing", "Puzzle", "Platform", "Sports", "Role-Playing", "Racing", "Shooter"],
    "Publisher": ["Nintendo"] * 10,
    "NA_Sales": [41.49, 29.08, 15.85, 11.27, 23.20, 11.38, 15.75, 9.00, 10.85, 26.93],
    "EU_Sales": [29.02, 3.58, 12.88, 8.89, 2.26, 9.23, 11.01, 6.00, 9.00, 0.63],
    "JP_Sales": [3.77, 6.81, 3.79, 10.22, 4.22, 6.50, 3.28, 7.20, 2.00, 0.28],
    "Other_Sales": [8.46, 0.77, 3.31, 1.00, 0.58, 2.90, 2.96, 1.00, 1.00, 0.47],
}

df = pd.DataFrame(data)
df["Global_Sales"] = df[["NA_Sales", "EU_Sales", "JP_Sales", "Other_Sales"]].sum(axis=1)

# Create SQLite in-memory database
conn = sqlite3.connect(":memory:")
df.to_sql("video_game_sales", conn, index=False, if_exists="replace")
```

```python
query1 = """
SELECT Name, Platform, Year, Genre, Publisher, Global_Sales
FROM video_game_sales
ORDER BY Global_Sales DESC
LIMIT 10;
"""
pd.read_sql_query(query1, conn)
```

```python
query2 = """
SELECT
  SUM(NA_Sales) AS Total_NA_Sales,
  SUM(EU_Sales) AS Total_EU_Sales,
  SUM(JP_Sales) AS Total_JP_Sales,
  SUM(Other_Sales) AS Total_Other_Sales,
  SUM(Global_Sales) AS Total_Global_Sales,
  ROUND(SUM(NA_Sales) / SUM(Global_Sales) * 100, 2) AS NA_Percentage,
  ROUND(SUM(EU_Sales) / SUM(Global_Sales) * 100, 2) AS EU_Percentage,
  ROUND(SUM(JP_Sales) / SUM(Global_Sales) * 100, 2) AS JP_Percentage,
  ROUND(SUM(Other_Sales) / SUM(Global_Sales) * 100, 2) AS Other_Percentage
FROM video_game_sales;
"""
pd.read_sql_query(query2, conn)
```

```python
query3 = """
WITH DecadeSales AS (
  SELECT
    CASE 
      WHEN Year BETWEEN 1980 AND 1989 THEN '1980s'
      WHEN Year BETWEEN 1990 AND 1999 THEN '1990s'
      WHEN Year BETWEEN 2000 AND 2009 THEN '2000s'
      WHEN Year BETWEEN 2010 AND 2020 THEN '2010s'
    END AS Decade,
    Publisher,
    SUM(Global_Sales) AS Total_Global_Sales
  FROM video_game_sales
  GROUP BY Decade, Publisher
),
Ranked AS (
  SELECT *,
         ROW_NUMBER() OVER (PARTITION BY Decade ORDER BY Total_Global_Sales DESC) as Rank
  FROM DecadeSales
)
SELECT Decade, Publisher, Total_Global_Sales
FROM Ranked
WHERE Rank <= 3;
"""
pd.read_sql_query(query3, conn)
```

```python
query4 = """
SELECT
  Platform,
  MIN(Year) AS First_Release_Year,
  MAX(Year) AS Last_Release_Year,
  COUNT(*) AS Total_Games,
  ROUND(AVG(Global_Sales), 2) AS Avg_Global_Sales,
  ROUND(SUM(Global_Sales), 2) AS Total_Global_Sales
FROM video_game_sales
GROUP BY Platform
ORDER BY Total_Global_Sales DESC;
"""
pd.read_sql_query(query4, conn)
```

```python
query5 = """
SELECT
  Genre,
  SUM(CASE WHEN Year BETWEEN 1980 AND 1999 THEN 1 ELSE 0 END) AS Games_1980_1999,
  SUM(CASE WHEN Year BETWEEN 1980 AND 1999 THEN Global_Sales ELSE 0 END) AS Sales_1980_1999,
  SUM(CASE WHEN Year BETWEEN 2000 AND 2009 THEN 1 ELSE 0 END) AS Games_2000_2009,
  SUM(CASE WHEN Year BETWEEN 2000 AND 2009 THEN Global_Sales ELSE 0 END) AS Sales_2000_2009,
  SUM(CASE WHEN Year BETWEEN 2010 AND 2020 THEN 1 ELSE 0 END) AS Games_2010_2020,
  SUM(CASE WHEN Year BETWEEN 2010 AND 2020 THEN Global_Sales ELSE 0 END) AS Sales_2010_2020
FROM video_game_sales
GROUP BY Genre
ORDER BY Genre;
"""
pd.read_sql_query(query5, conn)
```

```python
df_cleaned = df.drop_duplicates()
```

```python
def cleaning_report(before, after):
    print("Data Cleaning Summary Report:\n")
    print(f"Rows before cleaning: {before.shape[0]}")
    print(f"Rows after cleaning:  {after.shape[0]}")
    print("\nMissing values BEFORE:\n", before.isna().sum())
    print("\nMissing values AFTER:\n", after.isna().sum())
    print("\nDuplicates removed:", before.duplicated().sum())

cleaning_report(df, df_cleaned)
```

```python
grouped = df_cleaned.groupby(["Publisher", "Platform"])["Global_Sales"].agg(
    Mean_Sales="mean",
    Median_Sales="median",
    Max_Sales="max",
    Game_Count="count"
).sort_values(by="Mean_Sales", ascending=False)

print("Top 10 Publisher + Platform combinations by Mean Global Sales:")
display(grouped.head(10))

top_max_sales = grouped.sort_values(by="Max_Sales", ascending=False).head(5)
print("Top 5 Publisher + Platform combinations by Max Global Sales:")
display(top_max_sales)

platform_counts = df_cleaned.groupby("Publisher")["Platform"].nunique().sort_values(ascending=False)
print("Top 10 Publishers by Number of Platforms:")
display(platform_counts.head(10))

summary = df_cleaned.groupby("Publisher").agg(
    Unique_Platforms=("Platform", "nunique"),
    Total_Global_Sales=("Global_Sales", "sum"),
    Total_Games=("Name", "count")
).sort_values(by="Unique_Platforms", ascending=False)

print("Publisher Summary (by platform diversity):")
display(summary.head(10))
```

```python
import pandas as pd
import matplotlib.pyplot as plt


df_cleaned["Year"] = pd.to_numeric(df_cleaned["Year"], errors="coerce")


df_yearly = df_cleaned[(df_cleaned["Year"] >= 1980) & (df_cleaned["Year"] <= 2020)]


annual_sales = df_yearly.groupby("Year")["Global_Sales"].sum().reset_index()

print("Annual Global Sales (1980–2020):")
display(annual_sales)


plt.figure(figsize=(12, 6))
plt.plot(annual_sales["Year"], annual_sales["Global_Sales"], marker='o')
plt.title("Annual Global Video Game Sales (1980–2020)")
plt.xlabel("Year")
plt.ylabel("Global Sales (Millions)")
plt.grid(True)
plt.tight_layout()
plt.show()


peak_year = annual_sales[annual_sales["Global_Sales"] == annual_sales["Global_Sales"].max()]
print("Peak Year(s) for Global Sales:")
display(peak_year)

annual_sales["YoY_Growth_%"] = annual_sales["Global_Sales"].pct_change() * 100

print("Annual Sales with Year-over-Year Growth:")
display(annual_sales)


plt.figure(figsize=(12, 5))
plt.bar(annual_sales["Year"], annual_sales["YoY_Growth_%"], color='skyblue')
plt.axhline(0, color='gray', linestyle='--')
plt.title("Year-over-Year Growth in Global Sales")
plt.xlabel("Year")
plt.ylabel("Growth Rate (%)")
plt.tight_layout()
plt.show()

#
release_trend = df_yearly.groupby("Year")["Name"].count().reset_index(name="Games_Released")

print("Annual Game Release Count (Cyclical Trends):")
display(release_trend)


plt.figure(figsize=(12, 5))
plt.plot(release_trend["Year"], release_trend["Games_Released"], marker='o', color='green')
plt.title("Number of Games Released Per Year")
plt.xlabel("Year")
plt.ylabel("Game Count")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns

sales_columns = ["NA_Sales", "EU_Sales", "JP_Sales", "Other_Sales", "Global_Sales"]

correlation_matrix = df_cleaned[sales_columns].corr()
print("Correlation Coefficients Between Regional Sales:")
display(correlation_matrix)

plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", linewidths=0.5)
plt.title("Correlation Between Regional Sales")
plt.tight_layout()
plt.show()

Q1 = df_cleaned["Global_Sales"].quantile(0.25)
Q3 = df_cleaned["Global_Sales"].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df_cleaned[(df_cleaned["Global_Sales"] < lower_bound) | (df_cleaned["Global_Sales"] > upper_bound)]
print(f"Detected {len(outliers)} Outliers in Global Sales using IQR Method")
display(outliers[["Name", "Platform", "Global_Sales"]])

genre_stats = df_cleaned.groupby("Genre")["Global_Sales"].agg(
    Mean="mean",
    StdDev="std",
    Q1=lambda x: x.quantile(0.25),
    Median="median",
    Q3=lambda x: x.quantile(0.75),
    Max="max",
    Count="count"
).sort_values(by="Mean", ascending=False)

print("Genre-wise Statistical Summary:")
display(genre_stats)

sample = df_cleaned["Global_Sales"]
sample = sample[sample > 0]

shapiro_stat, shapiro_p = stats.shapiro(sample.sample(n=min(len(sample), 5000), random_state=1))
print("Normality Test (Shapiro-Wilk) for Global Sales:")
print(f"Test Statistic: {shapiro_stat:.4f}, p-value: {shapiro_p:.4f}")

if shapiro_p < 0.05:
    print("Distribution is likely NOT normal (reject H0).")
else:
    print("Distribution is likely normal (fail to reject H0).")

plt.figure(figsize=(10, 5))
sns.histplot(sample, bins=40, kde=True)
plt.title("Distribution of Global Sales")
plt.xlabel("Global Sales (Millions)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df_cleaned["Year"] = pd.to_numeric(df_cleaned["Year"], errors="coerce")
df_cleaned = df_cleaned[(df_cleaned["Year"] >= 1980) & (df_cleaned["Year"] <= 2020)]

publisher_share = df_cleaned.groupby("Publisher").agg(
    Total_Global_Sales=("Global_Sales", "sum"),
    Title_Count=("Name", "count")
).sort_values(by="Total_Global_Sales", ascending=False)

publisher_share["Sales_Share_%"] = 100 * publisher_share["Total_Global_Sales"] / publisher_share["Total_Global_Sales"].sum()
publisher_share["Title_Share_%"] = 100 * publisher_share["Title_Count"] / publisher_share["Title_Count"].sum()

display(publisher_share.head(10))

df_cleaned["Decade"] = pd.cut(df_cleaned["Year"], bins=[1979, 1989, 1999, 2009, 2020],
                              labels=["1980s", "1990s", "2000s", "2010s"])

platform_evolution = df_cleaned.groupby(["Decade", "Platform"])["Global_Sales"].sum().reset_index()
platform_pivot = platform_evolution.pivot(index="Platform", columns="Decade", values="Global_Sales").fillna(0)
display(platform_pivot.sort_values(by="2010s", ascending=False))

post_2010 = df_cleaned[df_cleaned["Year"] >= 2010]
pre_2010 = df_cleaned[df_cleaned["Year"] < 2010]

recent_sales = post_2010.groupby("Publisher")["Global_Sales"].sum().sort_values(ascending=False)
past_sales = pre_2010.groupby("Publisher")["Global_Sales"].sum()

emerging_declining = pd.DataFrame({
    "Pre_2010_Sales": past_sales,
    "Post_2010_Sales": recent_sales
}).fillna(0)

emerging_declining["Change_%"] = 100 * (emerging_declining["Post_2010_Sales"] - emerging_declining["Pre_2010_Sales"]) / (emerging_declining["Pre_2010_Sales"] + 1)

emerging = emerging_declining.sort_values(by="Change_%", ascending=False).head(5)
declining = emerging_declining.sort_values(by="Change_%", ascending=True).head(5)

print("Emerging Publishers (post-2010):")
display(emerging)

print("Declining Publishers (post-2010):")
display(declining)

sales_total = df_cleaned["Global_Sales"].sum()
market_shares = df_cleaned.groupby("Publisher")["Global_Sales"].sum() / sales_total
hhi = (market_shares ** 2).sum() * 10000

print(f"Herfindahl-Hirschman Index (HHI) for Publisher Market Concentration: {hhi:.2f}")
```
