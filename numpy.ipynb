```python
import numpy as np


salaries = np.array([
    68000, 72000, 75000, 80000, 85000, 87000, 90000, 95000, 98000, 100000,
    102000, 105000, 110000, 115000, 118000, 120000, 125000, 130000, 135000, 140000,
    142000, 145000, 150000, 155000, 160000, 165000, 170000, 175000, 180000, 185000,
    190000, 195000, 200000, 210000, 220000, 230000, 240000, 250000, 260000, 270000,
    280000, 290000, 300000, 320000, 340000, 360000, 380000, 400000, 420000, 450000
])

np.random.seed(0)
experience_levels = np.random.choice(
    ['EN', 'MD', 'SE'],
    size=len(salaries),
    p=[0.4, 0.4, 0.2]
)

senior_mask = experience_levels == 'SE'
entry_mask = experience_levels == 'EN'

senior_salaries = salaries[senior_mask]
entry_salaries = salaries[entry_mask]

avg_senior_salary = np.mean(senior_salaries)
avg_entry_salary = np.mean(entry_salaries)

percent_senior = (np.sum(senior_mask) / len(salaries)) * 100

print(f"Senior Salaries ({len(senior_salaries)} employees): {senior_salaries}")
print(f"Average Senior Salary: ${avg_senior_salary:,.2f}")
print(f"Average Entry-level Salary: ${avg_entry_salary:,.2f}")
print(f"Percentage of Senior Employees: {percent_senior:.2f}%")
```

```python
import numpy as np

work_year = np.array([
    2020, 2020, 2021, 2021, 2021, 2022, 2022, 2022, 2023, 2023, 
    2023, 2023, 2024, 2024, 2024, 2024, 2024, 2025, 2025, 2025
])

salary_in_usd = np.array([
    70000, 80000, 75000, 77000, 85000, 90000, 95000, 97000, 100000, 
    105000, 110000, 115000, 120000, 125000, 130000, 135000, 140000, 
    145000, 150000, 155000
])

unique_years = np.unique(work_year)
mean_salaries = []

for year in unique_years:
    mean_salary = np.mean(salary_in_usd[work_year == year])
    mean_salaries.append(mean_salary)

mean_salaries = np.array(mean_salaries)

result = np.column_stack((unique_years, mean_salaries))

for year, salary in result:
    print(f"Year: {int(year)} | Mean Salary: ${int(salary):,}")
```

```python
import numpy as np
import pandas as pd

salary_in_usd = np.array([
    70000, 80000, 75000, 77000, 85000, 90000, 95000, 97000, 100000, 
    105000, 110000, 115000, 120000, 125000, 130000, 135000, 140000, 
    145000, 150000, 155000
])

company_size = np.array([
    'S', 'M', 'L', 'L', 'M', 'S', 'S', 'M', 'L', 'L',
    'S', 'M', 'L', 'L', 'S', 'S', 'M', 'L', 'M', 'S'
])

salaries_S = salary_in_usd[company_size == 'S']
salaries_M = salary_in_usd[company_size == 'M']
salaries_L = salary_in_usd[company_size == 'L']

def describe(data):
    return {
        "mean": np.mean(data),
        "median": np.median(data),
        "std": np.std(data),
        "min": np.min(data),
        "max": np.max(data)
    }

stats_S = describe(salaries_S)
stats_M = describe(salaries_M)
stats_L = describe(salaries_L)

print("Small Company Stats:", stats_S)
print("Medium Company Stats:", stats_M)
print("Large Company Stats:", stats_L)

variances = {
    'S': np.var(salaries_S),
    'M': np.var(salaries_M),
    'L': np.var(salaries_L)
}

max_variance_size = max(variances, key=variances.get)
print("\nCompany size with highest salary variance:", max_variance_size)

percentiles_S = np.percentile(salaries_S, [25, 50, 75])
percentiles_M = np.percentile(salaries_M, [25, 50, 75])
percentiles_L = np.percentile(salaries_L, [25, 50, 75])

percentile_array = np.array([percentiles_S, percentiles_M, percentiles_L])

print("\nPercentiles (rows = S, M, L | cols = 25th, 50th, 75th):\n", percentile_array)

percentile_df = pd.DataFrame(
    percentile_array,
    index=["Small", "Medium", "Large"],
    columns=["25th", "50th", "75th"]
)
print("\nPretty Percentile Table:\n")
display(percentile_df)
```

```python
import numpy as np

remote_ratio = np.array([
    100, 0, 50, 100, 0, 50, 100, 0, 50, 100,
    0, 50, 100, 0, 50, 100, 0, 50, 100, 0
])

salary_in_usd = np.array([
    70000, 80000, 75000, 77000, 85000, 90000, 95000, 97000, 100000,
    105000, 110000, 115000, 120000, 125000, 130000, 135000, 140000,
    145000, 150000, 155000
])

fully_remote_salaries = salary_in_usd[remote_ratio == 100]
hybrid_salaries = salary_in_usd[remote_ratio == 50]
onsite_salaries = salary_in_usd[remote_ratio == 0]

avg_fully_remote = np.mean(fully_remote_salaries)
avg_hybrid = np.mean(hybrid_salaries)
avg_onsite = np.mean(onsite_salaries)

print("Average Salaries:")
print(f"Fully Remote (100%): ${avg_fully_remote:.2f}")
print(f"Hybrid (50%): ${avg_hybrid:.2f}")
print(f"On-site (0%): ${avg_onsite:.2f}")

mean_salary = np.mean(salary_in_usd)
std_salary = np.std(salary_in_usd)
normalized_salaries = (salary_in_usd - mean_salary) / std_salary

correlation = np.corrcoef(remote_ratio, salary_in_usd)[0,1]

print(f"\nCorrelation between remote work ratio and salary: {correlation:.3f}")
```
